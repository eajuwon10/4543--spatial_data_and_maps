## Q4. File I/O
Write Python code that saves `Hello World` to a file called `hello.txt`, then reads it back and prints the contents.

```python
# Write to the file
with open("hello.txt", "w") as file:
    file.write("Hello World")

# Read the file and print its contents
with open("hello.txt", "r") as file:
    contents = file.read()
    print(contents)
```

## Q6. Plotting
Create a line plot of $y = x^2$ for $x$ values from 0 to 10.

```python
!pip install matplotlib
```

```python
import matplotlib.pyplot as plt

x = list(range(0, 11))        # values 0 through 10
y = [i**2 for i in x]         # y = x^2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y = x^2")
plt.title("Line Plot of y = x^2")
plt.show()
```
