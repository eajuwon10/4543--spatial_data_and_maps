# 📈 05 - Plotting Basics

Visualization is a core part of data science.  
In this notebook you will learn:
- Setting up Matplotlib for inline plotting
- Creating simple line, scatter, and bar plots
- Customizing titles, labels, and legends
- Why visuals matter in data storytelling


## 1. Setup

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Line Plot

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)")
plt.title("Line Plot Example")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.show()
```

✅ **Your Turn**: Plot `cos(x)` on the same graph with a different color and label.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.linspace(0, 2*np.pi, 400)

# Plot cos(x) with a different color and label
plt.plot(x, np.cos(x), label="cos(x)", color="red")

# Add labels and legend
plt.xlabel("x")
plt.ylabel("y")
plt.title("cos(x)")
plt.legend()

plt.show()

```

## 3. Scatter Plot

```python
x = np.random.rand(50)
y = np.random.rand(50)
sizes = 300 * np.random.rand(50)
colors = np.random.rand(50)

plt.scatter(x, y, s=sizes, c=colors, alpha=0.6, cmap='viridis')
plt.title("Scatter Plot Example")
plt.xlabel("x")
plt.ylabel("y")
plt.colorbar(label="color scale")
plt.show()
```

✅ **Your Turn**: Generate two sets of random points and plot them in different colors on the same scatter plot.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate two sets of random points
x1 = np.random.randn(100)
y1 = np.random.randn(100)

x2 = np.random.randn(100) + 3   # shifted so they don't overlap completely
y2 = np.random.randn(100) + 3

# Create scatter plot
plt.scatter(x1, y1, color="blue", label="Group 1")
plt.scatter(x2, y2, color="red", label="Group 2")

plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Two Sets of Random Points")
plt.legend()
plt.show()
```

## 4. Bar Plot

```python
categories = ["A", "B", "C", "D"]
values = [5, 7, 3, 8]

plt.bar(categories, values)
plt.title("Bar Plot Example")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

✅ **Your Turn**: Create your own bar chart with at least 4 categories (e.g., favorite foods, courses, or hobbies).

```python
import matplotlib.pyplot as plt

# Data
categories = ["Apples", "Bananas", "Cherries", "Dates"]
values = [10, 15, 7, 12]

# Create bar chart
plt.bar(categories, values, color=["red", "yellow", "purple", "brown"])

plt.xlabel("Fruit")
plt.ylabel("Quantity")
plt.title("Fruit Quantities")
plt.show()
```

## 5. Customizing Plots

```python
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin(x)", linestyle="--", color="blue")
plt.plot(x, y2, label="cos(x)", linestyle=":", color="red")
plt.title("Customized Line Plot")
plt.xlabel("x (radians)")
plt.ylabel("value")
plt.legend()
plt.grid(True)
plt.show()
```

✅ **Your Turn**: Customize your own plot with labels, legend, and grid lines.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create data
x = np.linspace(0, 10, 200)
y1 = np.sin(x)
y2 = np.cos(x)

# Plot
plt.plot(x, y1, label="sin(x)", color="blue")
plt.plot(x, y2, label="cos(x)", color="red")

# Labels and title
plt.xlabel("X values")
plt.ylabel("Function value")
plt.title("Sine and Cosine Functions")

# Legend
plt.legend()

# Grid lines
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```

---
### Summary
- `%matplotlib inline` makes plots display inside the notebook.
- Line, scatter, and bar plots are the most common basics.
- Titles, labels, legends, and grids improve readability.
- Visuals tell stories about data more effectively than raw numbers.
