## ✅ Learning Goals

```python
# 🧠 Variables and Types
x = 10
y = 3.5
z = 'hello'
print(type(x), type(y), type(z))
```

```python
# ➕ Arithmetic Operations
print('Add:', x + y)
print('Divide:', x / y)
print('Floor Divide:', x // y)
print('Exponent:', x ** 2)
```
